```python
import numpy as np
import pandas as pd
data=np.array(['a','b','c','d'])
s=pd.Series(data)
print(s)
```

```python
data=np.array(["a","b","c","d"])
s=pd.Series(data,index=[100,101,102,103])
print(s)
```

```python
#Dict Series
data={"a":1.1,"b":2.2,"c":3,"d":"age"}
s=pd.Series(data)
print(s)
```

```python
data={"a":1,"b":2,"c":3}
s=pd.Series(data,index=["a","d","b","c"])
print(s)
```

```python
#INDEXING AND SLICING
s=pd.Series([1,2,3,4,5])
i=s[1]
print(i)

print(s[0:4])
```

```python
s=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s['d'])
print(s['b'])
print(s['b':'e'])
print(s[['a','c','e']])
```

```python
#Creating an empty DataFrame
df=pd.DataFrame()
print(df)
```

```python
data=[1,2,3,4,5]
df=pd.DataFrame(data)
print(df)
```

```python
#Creating a DataFrame from a Dict of list
data={'Name':['a','b','c','d'],
     'Age':[20,21,22,23]}
df=pd.DataFrame(data)
print(df)
```

```python
d={'one':pd.Series([1,2,3],index=['a','b','c']),
'Two':pd.Series([1,2,3,4],index=['a','b','c','d'])}
df=pd.DataFrame(d)
print(df)
```

```python
#Select a single column
d={'one':pd.Series([1,2,3],index=['a','b','c']),
'Two':pd.Series([1,2,3,4],index=['a','b','c','d'])}
df=pd.DataFrame(d)
print(df['one'])
print(df[['one','Two']])
```

```python
d={'One':pd.Series([1,2,3],index=['A','B','C']),
'Two':pd.Series([1,2,3,4],index=['A','B','C','D'])}
df=pd.DataFrame(d)
#ADDING A NEW COLUMN BY PASSING SERIES
df['Three']=pd.Series([10,20,30],index=['A','B','C'])
df
```

```python
#Creating a table using random values
from numpy.random import randn
df=pd.DataFrame(randn(5,5,),index='A B C D E'.split(),columns='V W X Y Z'.split())
df
```

```python
from numpy.random import randn
df=pd.DataFrame(randn(5,5,),index='A B C D E'.split(),columns='V W X Y Z'.split())
print(df)
#Take the rows of the table 
r1=df.loc['B']      #Lable Based(loc)
print("\n-------------\n",r1)
r2=df.iloc[2]
print(r2)

rc1=df.loc[['A','B'],['W','Y']]
rc1
```
