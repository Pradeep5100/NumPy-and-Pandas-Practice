```python
# Q1: Create a 2D array of shape (2, 3) with values from 1 to 6.
a=np.array([[1,2,3],[4,5,6]])
print(a)
# Q2: Create a 2D array with all values set to 9 of shape (3, 3).
a=np.array([[9,9,9],[9,9,9],[9,9,9]])
print(a)
```

```python
# Q1: Create any array and print its type using type().
a=np.array([2,3,4,5])
print(type(a))
```

```python
# Q1: Check how many dimensions a 2D array has.
a=np.array([[1,2,3],[4,5,6]])
print(a.ndim)
a=np.array([2,3,4,5])
print(a.ndim)
```

```python
# Q1: Find the total number of elements in a 2D array.
a=np.array([[1,2,3],[4,5,6]])
print(a.size)
```

```python
# Q1: Create a 4x4 identity matrix using np.eye().
a=np.array([[1,2,3],[4,5,6]])
print(a.reshape(3,2))
print(np.eye(4,dtype=int))
```

```python
# Q1: Create an array with numbers from 0 to 10 (excluding 10) with step size 2.
arr=np.arange(1,10)
print(arr)
```

```python
# Q1: Generate 5 numbers between 0 and 1 (inclusive).
x=np.linspace(0,1,5)
print(x)
```

```python
# Q1: Generate a 2x2 array with random values between 0 and 1.
x=np.random.rand(2,2)
print(x)
# Q1: Create a 3x3 matrix with random numbers from a standard normal distribution.
x=np.random.randn(3,3)
print(x)
# Q1: Generate a 1D array with 5 random integers between 1 and 100.
x=np.random.randint(1,100,5)
print(x)
```

```python
# Q1: Find the maximum value in the array [10, 20, 5, 45].
a=np.array([[1,2,3],[4,5,6]])
print(a.max())
```

```python
# Q1: Find the minimum value in the array [10, 20, 5, 45].
a=np.array([10, 20, 5, 45])
print(a.min())
```

```python
# Q1: Find the index of the maximum element in the array [3, 7, 2, 10, 6].
a=np.array([3, 7, 2, 10, 6])
print(a.argmax())
print(a.argmin())
```

```python
# 17.  NumPy Indexing and Selection (1D, 2D)
# Q1: Create a 1D array [10, 20, 30, 40, 50] and access the third element.
x=np.array([10, 20, 30, 40, 50])
a=x[2]
print(a)
# Q2: Create a 2D array [[1, 2], [3, 4], [5, 6]] and access the element 4.
x=np.array([[1, 2], [3, 4], [5, 6]])
b=x[1][1]
print(b)
```

```python
# Q1: Join two 1D arrays [1, 2] and [3, 4]. numpy.split()
x=np.array([1, 2])
y=np.array([3, 4])
print(np.concatenate((x,y),axis=0))
```

```python
# Q1: Split [10, 20, 30, 40, 50, 60] into 3 equal parts.
x=np.array([10, 20, 30, 40, 50, 60])
print(np.split(x,3))
```

```python
# Q1: Horizontally split a 2D array:
arr=np.array([[1,2],[3,4]])
print(np.hsplit(arr,2))
# Q1: Vertically split the same 2D array into 2 parts.
print(np.vsplit(arr,2))
```

```python
# Q1: Resize [1, 2, 3, 4] into 2x2.
arr=np.array([1, 2, 3, 4])
print(np.resize(arr,[2,2]))
# Q2: Resize [1, 2, 3, 4, 5, 6] into 3x2.
arr=np.array([1, 2, 3, 4, 5, 6])
print(np.resize(arr,[3,2]))
```

```python
# Q1: Append [4, 5] to [1, 2, 3].
arr=np.array([1, 2, 3])
print(np.append(arr,[4,5]))
```

```python
# Q1: Insert 100 at index 1 in [10, 20, 30].
arr=np.array([10, 20, 30])
print(np.insert(arr,1,100))
```

```python
# Q1: Delete the second element from [1, 2, 3].
arr=np.array([1, 2, 3])
print(np.delete(arr,1))
```
